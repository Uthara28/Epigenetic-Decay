```python

fig, ax = plt.subplots(3, 4, figsize=(18, 9),
                        gridspec_kw={"height_ratios": [1, 1, 0.15]})
#fig.suptitle(
 #   r"Evolved transgenerational memory and neutral trait across values of traits contributing to phenotype $\mathit{n}$",
 #   fontsize=22)
 
fig.subplots_adjust(top=0.88, bottom=0.11, hspace=0.35, wspace=0.25)

plot_no = ['i)', 'ii)', 'iii)', 'iv)']

cmap = plt.cm.viridis.copy()
cmap.set_bad(color='gray')
norm = mcolors.Normalize(vmin=0, vmax=1)

dims = [1, 2, 5, 10]
pcms = []

for col, i in enumerate(dims):
    df_epi = epi_obj[f"epi_{i}"]

    # --- Transgenerational memory (top row) ---
    pivoted_epi = (
        df_epi.pivot_table(index='m', columns='rho', values='epi_mem_p')
        .sort_index()
        .sort_index(axis=1)
    )

    pcm1 = ax[0, col].pcolormesh(
        pivoted_epi.columns,
        pivoted_epi.index,
        pivoted_epi.values,
        cmap=cmap,
        norm =norm,
        shading='auto'
    )
    pcms.append(pcm1)

    ax[0, col].set_ylabel('', fontsize=15)
    ax[0, col].set_xlabel(r'$$', fontsize=15)
    ax[0, col].set_xticks(np.arange(-1, 1.1, 0.5))
    ax[0, col].set_yticks(np.arange(0.50, 1.0, 0.1))
    ax[0, col].tick_params(axis='both', labelsize=10)
    ax[0, col].set_xlim(0,1)
    
    # --- Neutral trait (middle row) ---
    pivoted_neutral = (
        df_epi.pivot_table(index='m', columns='rho', values='neutral_trait_p')
        .sort_index()
        .sort_index(axis=1)
    )

    pcm2 = ax[1, col].pcolormesh(
        pivoted_neutral.columns,
        pivoted_neutral.index,
        pivoted_neutral.values,
        cmap=cmap,
        norm =norm,
        shading='auto'
    )
    pcms.append(pcm2)

    ax[1, col].set_ylabel('m', fontsize=15)
    ax[1, col].set_xlabel(r'$\rho$', fontsize=15)
    ax[1, col].set_xticks(np.arange(-1, 1.1, 0.5))
    ax[1, col].set_yticks(np.arange(0.50, 1.0, 0.1))
    ax[1, col].tick_params(axis='both', labelsize=10)
    ax[0, col].set_xlim(0,1)
    # --- Column header: n = 1, 2, 5, 10 (top row title) ---
    ax[0, col].set_title(
        rf"{plot_no[col]} $n = {i}$",
        fontsize=16,
        style="italic"
    )
row_headers = ['(a)', '(b)']
# Add row headers on the left side using stored axes
for row, header in enumerate(row_headers):
    ax[row][0].text(-0.32, 0.5, header,
                      transform=ax[row][0].transAxes,
                      rotation=90,
                      fontsize=16,
                      fontweight='bold',
                      va='center',
                      ha='center',
                      zorder=10)


# --- Third row used as colorbar row ---
# Turn off plotting in the bottom row
for col in range(4):
    ax[2, col].set_axis_off()

# Create a single colorbar axis spanning the third row

gs = ax[0, 0].get_gridspec()  # reuse the same gridspec
cbar_ax = fig.add_subplot(gs[2, 1:3])  # third row, all columns

cbar = fig.colorbar(
    pcms[-1],          # reference the last plotted mesh (neutral trait)
    cax=cbar_ax,
    orientation='horizontal',
    pad=0.05,
    shrink=0.8
)
cbar.set_label('Evolved phenotypic memory and neutral trait', fontsize=14)
cbar.set_ticks([0, 0.25, 0.5, 0.75, 1.0])

plt.tight_layout()
plt.savefig('/home/usriniva/Desktop/Phd/Epigenetics/Simulations/New_Hanna_eqns/Evolve_Transgenerational_memory/Plot_results/plot_report/figures_svg/alldim_finegrain.png',
            transparent=True)
plt.show()
```

```python

fig, ax = plt.subplots(3, 4, figsize=(18, 9),
                        gridspec_kw={"height_ratios": [1, 1, 0.15]})
#fig.suptitle(
 #   r"Evolved transgenerational memory and neutral trait across values of traits contributing to phenotype $\mathit{n}$",
 #   fontsize=22)
 
fig.subplots_adjust(top=0.88, bottom=0.11, hspace=0.35, wspace=0.25)

plot_no = ['i)', 'ii)', 'iii)', 'iv)']

cmap = plt.cm.viridis.copy()
cmap.set_bad(color='gray')
norm = mcolors.Normalize(vmin=0, vmax=1)

dims = [1, 2, 5, 10]
pcms = []

for col, i in enumerate(dims):
    df_epi = epi_obj[f"epi_{i}"]

    # --- Transgenerational memory (top row) ---
    pivoted_epi = (
        df_epi.pivot_table(index='m', columns='rho', values='epi_mem_p')
        .sort_index()
        .sort_index(axis=1)
    )

    pcm1 = ax[0, col].pcolormesh(
        pivoted_epi.columns,
        pivoted_epi.index,
        pivoted_epi.values,
        cmap=cmap,
        norm =norm,
        shading='auto'
    )
    pcms.append(pcm1)

    ax[0, col].set_ylabel('m', fontsize=15)
    ax[0, col].set_xlabel(r'$\rho$', fontsize=15)
    ax[0, col].set_xticks(np.arange(-1, 1.1, 0.5))
    ax[0, col].set_yticks(np.arange(0.50, 1.0, 0.1))
    ax[0, col].tick_params(axis='both', labelsize=10)

    # --- Neutral trait (middle row) ---
    pivoted_neutral = (
        df_epi.pivot_table(index='m', columns='rho', values='neutral_trait_p')
        .sort_index()
        .sort_index(axis=1)
    )

    pcm2 = ax[1, col].pcolormesh(
        pivoted_neutral.columns,
        pivoted_neutral.index,
        pivoted_neutral.values,
        cmap=cmap,
        norm =norm,
        shading='auto'
    )
    pcms.append(pcm2)

    ax[1, col].set_ylabel('m', fontsize=15)
    ax[1, col].set_xlabel(r'$\rho$', fontsize=15)
    ax[1, col].set_xticks(np.arange(-1, 1.1, 0.5))
    ax[1, col].set_yticks(np.arange(0.50, 1.0, 0.1))
    ax[1, col].tick_params(axis='both', labelsize=10)

    # --- Column header: n = 1, 2, 5, 10 (top row title) ---
    ax[0, col].set_title(
        rf"{plot_no[col]} $n = {i}$",
        fontsize=16,
        style="italic"
    )

# --- Third row used as colorbar row ---
# Turn off plotting in the bottom row
for col in range(4):
    ax[2, col].set_axis_off()

# Create a single colorbar axis spanning the third row

gs = ax[0, 0].get_gridspec()  # reuse the same gridspec
cbar_ax = fig.add_subplot(gs[2, 1:3])  # third row, all columns

cbar = fig.colorbar(
    pcms[-1],          # reference the last plotted mesh (neutral trait)
    cax=cbar_ax,
    orientation='horizontal',
    pad=0.05,
    shrink=0.8
)
cbar.set_label('Evolved transgenerational memory and neutral trait', fontsize=14)
cbar.set_ticks([0, 0.25, 0.5, 0.75, 1.0])

plt.tight_layout()
plt.show()
```

```python

# Your existing preamble ...
fig, ax = plt.subplots(len(dims), 3, figsize=(30, 8 * len(dims)))

# Turn 1D ax into 2D if needed
if len(dims) == 1:
    ax = np.expand_dims(ax, axis=0)

# Generate row headers
row_headers = [f'No. traits: {i}' for i in dims]

for row, i in enumerate(dims):
    df_epi = epi_obj[f"epi_{i}"]
    df_gen = onlygen_obj[f"gen_{i}"]

    cmap1 = plt.cm.copper.copy()
    cmap1.set_bad(color='gray')

    pivoted_fitness = df_epi.pivot_table(index='m', columns='rho', values='fitness').sort_index().sort_index(axis=1)
    pivoted_gen     = df_gen.pivot_table(index='m', columns='rho', values='fitness').sort_index().sort_index(axis=1)

    norm = mcolors.Normalize(vmin=0, vmax=1)

    # Col 0: EPI fitness
    pcm1 = ax[row, 0].pcolormesh(pivoted_fitness.columns, pivoted_fitness.index, pivoted_fitness.values,
                                  cmap=cmap1, norm=norm, shading='auto')
    cbar0 = fig.colorbar(pcm1, ax=ax[row, 0])
    cbar0.set_label('Fitness (W) with TM', fontsize=axis_label_size)
    cbar0.ax.tick_params(labelsize=size)
    ax[row, 0].set_ylabel('mean time in env 1', fontsize=axis_label_size)
    ax[row, 0].set_xlabel('autocorrelation', fontsize=axis_label_size)

    # Col 1: GEN fitness
    pcm2 = ax[row, 1].pcolormesh(pivoted_gen.columns, pivoted_gen.index, pivoted_gen.values,
                                  cmap=cmap1, norm=norm, shading='auto')
    cbar1 = fig.colorbar(pcm2, ax=ax[row, 1])
    cbar1.set_label('Fitness without TM', fontsize=axis_label_size)
    cbar1.ax.tick_params(labelsize=size)
    ax[row, 1].set_ylabel('mean time in env 1', fontsize=axis_label_size)
    ax[row, 1].set_xlabel('autocorrelation', fontsize=axis_label_size)

    # Col 2: Difference
    pivoted_diff = pivoted_fitness - pivoted_gen
    norm_diff = mcolors.SymLogNorm(linthresh=0.03, linscale=0.03, vmin=-1.0, vmax=1.0)

    pcm_diff = ax[row, 2].pcolormesh(pivoted_diff.columns, pivoted_diff.index, pivoted_diff.values,
                                     cmap='BrBG', norm=norm_diff, shading='auto')
    cbar2 = fig.colorbar(pcm_diff, ax=ax[row, 2])
    cbar2.set_label('ΔFitness (TM - no TM)', fontsize=axis_label_size)
    cbar2.ax.tick_params(labelsize=size)
    ax[row, 2].set_ylabel('mean time in env 1', fontsize=axis_label_size)
    ax[row, 2].set_xlabel('autocorrelation', fontsize=axis_label_size)


# Column headers on FIRST ROW only (top row)
col_headers = ['Fitness with TM', 'Fitness without TM', 'ΔFitness']
for col, label in enumerate(col_headers):
    ax[0, col].set_title(label, fontsize=16, fontweight='bold', pad=20)


# Row headers using centered text on the LEFT COLUMN of each row
for row, header in enumerate(row_headers):
    ax[row, 0].set_title('', fontsize=16, fontweight='bold')  # clear any old title
    ax[row, 0].text(-0.25, 0.5, header,
                    transform=ax[row, 0].transAxes,
                    rotation=90,
                    fontsize=16,
                    fontweight='bold',
                    va='center',
                    ha='center',
                    bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='gray', alpha=0.8),
                    zorder=10)

# Adjust spacing between rows
fig.subplots_adjust(hspace=0.2)  # increase vertical space between rows
```
